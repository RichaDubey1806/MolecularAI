```python
x = 2
y = 6

w = 0
learning_rate = 0.1

for step in range(20):

    # Prediction
    prediction = w * x

    # Error
    error = y - prediction

    # Loss
    loss = error ** 2

    # Gradient
    gradient = -2 * x * error

    # Update
    w = w - learning_rate * gradient

    print(
        "Step:", step,
        "w:", round(w, 4),
        "loss:", round(loss, 4)
    )
```

```python
x = [1, 2, 3, 4]
y = [3, 6, 9, 12]

w = 0
learning_rate = 0.01

for step in range(1000):

    total_loss = 0
    gradient = 0

    for i in range(len(x)):

        prediction = w * x[i]

        error = y[i] - prediction

        loss = error ** 2

        total_loss += loss

        gradient += -2 * x[i] * error

    w = w - learning_rate * gradient

    if step % 100 == 0:
        print(
            "Step:", step,
            "w:", round(w, 4),
            "loss:", round(total_loss, 4)
        )
```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 8, 11, 14, 17])

w = 0
b = 0

learning_rate = 0.01

for step in range(1000):

    prediction = w * x + b

    error = y - prediction

    loss = np.mean(error ** 2)

    gradient_w = -2 * np.mean(x * error)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

    if step % 100 == 0:
        print(
            "Step:", step,
            "w:", round(w, 4),
            "b:", round(b, 4),
            "loss:", round(loss, 4)
        )
```

```python
import matplotlib.pyplot as plt

predictions = w * x + b

plt.scatter(x, y, label="Actual data")
plt.plot(x, predictions, label="Learned line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Our First Learned Model")
plt.legend()
plt.show()
```

```python
import numpy as np

X = np.array([
    [1, 5],
    [2, 6],
    [3, 7],
    [4, 8]
])

y = np.array([115, 145, 175, 205])

w = np.array([0.0, 0.0])
b = 0.0

learning_rate = 0.01

for step in range(1000):

    prediction = X @ w + b

    error = y - prediction

    loss = np.mean(error ** 2)

    gradient_w = -2 * (X.T @ error) / len(X)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

    if step % 100 == 0:
        print(
            "Step:", step,
            "w:", w,
            "b:", round(b, 4),
            "loss:", round(loss, 4)
        )
```

```python
import numpy as np

X = np.array([
    [1, 2],
    [2, 5],
    [3, 1],
    [4, 7],
    [5, 3]
])

# True relationship:
# y = 10*x1 + 20*x2 + 5

y = 10 * X[:, 0] + 20 * X[:, 1] + 5

w = np.array([0.0, 0.0])
b = 0.0

learning_rate = 0.01

for step in range(2000):

    prediction = X @ w + b
    error = y - prediction

    loss = np.mean(error ** 2)

    gradient_w = -2 * (X.T @ error) / len(X)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

    if step % 200 == 0:
        print(
            "Step:", step,
            "w:", np.round(w, 4),
            "b:", round(b, 4),
            "loss:", round(loss, 6)
        )
```
