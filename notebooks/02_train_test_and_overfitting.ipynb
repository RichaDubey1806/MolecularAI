```python
import numpy as np
import matplotlib.pyplot as plt

x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([2, 4, 6, 8, 10])

x_test = np.array([6, 7])
y_test = np.array([12, 14])

# A deliberately complicated model
x_curve = np.linspace(1, 7, 100)

y_curve = (
    2 * x_curve
    + 0.8 * (x_curve - 3) ** 2
    - 0.3 * (x_curve - 3) ** 3
)

plt.scatter(x_train, y_train, label="Training data")
plt.scatter(x_test, y_test, label="Test data")

plt.plot(x_curve, y_curve, label="Complicated model")

plt.xlabel("x")
plt.ylabel("y")
plt.title("A Model That Can Overfit")
plt.legend()

plt.show()
```

```python
# Compare a simple model with a complicated model

x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([2, 4, 6, 8, 10])

x_test = np.array([6, 7])
y_test = np.array([12, 14])

# Simple model
simple_prediction_train = 2 * x_train
simple_prediction_test = 2 * x_test

simple_train_error = np.mean((y_train - simple_prediction_train) ** 2)
simple_test_error = np.mean((y_test - simple_prediction_test) ** 2)

print("Simple model:")
print("Training error:", simple_train_error)
print("Test error:", simple_test_error)
```

```python
x_train = np.array([1, 2, 3, 4, 5])
y_train = np.array([2.1, 3.8, 6.2, 7.9, 10.3])

x_test = np.array([6, 7])
y_test = np.array([12.1, 13.9])

print("Training:", list(zip(x_train, y_train)))
print("Test:", list(zip(x_test, y_test)))
```

```python
# Simple linear model: y = wx + b

w = 0.0
b = 0.0
learning_rate = 0.01

for step in range(1000):
    prediction = w * x_train + b
    error = y_train - prediction

    gradient_w = -2 * np.mean(x_train * error)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

train_prediction = w * x_train + b
test_prediction = w * x_test + b

train_error = np.mean((y_train - train_prediction) ** 2)
test_error = np.mean((y_test - test_prediction) ** 2)

print("w =", w)
print("b =", b)
print("Training error =", train_error)
print("Test error =", test_error)
```

```python
# Create polynomial features

X_train_poly = np.column_stack([
    x_train,
    x_train**2,
    x_train**3,
    x_train**4
])

print(X_train_poly)
print("Shape:", X_train_poly.shape)
```

```python
w = np.zeros(4)
b = 0.0

learning_rate = 0.00001

for step in range(10000):
    prediction = X_train_poly @ w + b
    error = y_train - prediction

    gradient_w = -2 * (X_train_poly.T @ error) / len(x_train)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

train_prediction = X_train_poly @ w + b

train_error = np.mean((y_train - train_prediction) ** 2)

print("Weights:", w)
print("Bias:", b)
print("Training error:", train_error)
```

```python
X_train_poly_scaled = np.column_stack([
    x_train / 5,
    (x_train / 5)**2,
    (x_train / 5)**3,
    (x_train / 5)**4
])

print(X_train_poly_scaled)
```

```python
w = np.zeros(4)
b = 0.0

learning_rate = 0.01

for step in range(10000):
    prediction = X_train_poly_scaled @ w + b
    error = y_train - prediction

    gradient_w = -2 * (X_train_poly_scaled.T @ error) / len(x_train)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

train_prediction = X_train_poly_scaled @ w + b

train_error = np.mean((y_train - train_prediction) ** 2)

print("Weights:", w)
print("Bias:", b)
print("Training error:", train_error)
```

```python
X_test_poly_scaled = np.column_stack([
    x_test / 5,
    (x_test / 5)**2,
    (x_test / 5)**3,
    (x_test / 5)**4
])

test_prediction = X_test_poly_scaled @ w + b

test_error = np.mean((y_test - test_prediction) ** 2)

print("Actual:", y_test)
print("Predicted:", test_prediction)
print("Test error:", test_error)
```
