```python
from rdkit import Chem

print("RDKit imported successfully!")
```

```python
smiles = "CCO"

molecule = Chem.MolFromSmiles(smiles)

print("SMILES:", smiles)
print("Molecule:", molecule)
```

```python
print("Number of atoms:", molecule.GetNumAtoms())

for atom in molecule.GetAtoms():
    print(
        "Atom:",
        atom.GetIdx(),
        "Element:",
        atom.GetSymbol()
    )
```

```python
molecule_with_h = Chem.AddHs(molecule)

print("Atoms before adding H:", molecule.GetNumAtoms())
print("Atoms after adding H:", molecule_with_h.GetNumAtoms())

for atom in molecule_with_h.GetAtoms():
    print(
        "Atom:",
        atom.GetIdx(),
        "Element:",
        atom.GetSymbol()
    )
```

```python
print("Number of bonds:", molecule_with_h.GetNumBonds())

for bond in molecule_with_h.GetBonds():
    print(
        "Bond:",
        bond.GetBeginAtomIdx(),
        "—",
        bond.GetEndAtomIdx(),
        "Type:",
        bond.GetBondType()
    )
```

```python
from rdkit.Chem import rdMolDescriptors

formula = rdMolDescriptors.CalcMolFormula(molecule_with_h)

print("Molecular formula:", formula)
```

```python
molecular_weight = rdMolDescriptors.CalcExactMolWt(molecule)

print("Molecular weight:", molecular_weight)
```

```python
hbd = rdMolDescriptors.CalcNumHBD(molecule)
hba = rdMolDescriptors.CalcNumHBA(molecule)

print("Hydrogen bond donors:", hbd)
print("Hydrogen bond acceptors:", hba)
```

```python
rotatable_bonds = rdMolDescriptors.CalcNumRotatableBonds(molecule)

print("Rotatable bonds:", rotatable_bonds)
```

```python
tpsa = rdMolDescriptors.CalcTPSA(molecule)

print("TPSA:", tpsa)
```

```python
logp = Chem.Crippen.MolLogP(molecule)

print("LogP:", logp)
```

```python
def calculate_features(smiles):
    mol = Chem.MolFromSmiles(smiles)

    molecular_weight = rdMolDescriptors.CalcExactMolWt(mol)
    hbd = rdMolDescriptors.CalcNumHBD(mol)
    hba = rdMolDescriptors.CalcNumHBA(mol)
    rotatable_bonds = rdMolDescriptors.CalcNumRotatableBonds(mol)
    tpsa = rdMolDescriptors.CalcTPSA(mol)
    logp = Crippen.MolLogP(mol)

    return [
        molecular_weight,
        hbd,
        hba,
        rotatable_bonds,
        tpsa,
        logp
    ]
```

```python
ethanol_features = calculate_features("CCO")

print("Ethanol features:")
print(ethanol_features)
```

```python
all_features = []

for smiles in smiles_list:
    features = calculate_features(smiles)
    all_features.append(features)

print("Feature matrix:")
print(all_features)
```

```python
X = np.array(all_features)

print("X:")
print(X)

print("Shape:", X.shape)
```
